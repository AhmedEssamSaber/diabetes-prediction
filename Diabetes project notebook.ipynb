# <span style="color:rgb(102, 133, 190); font-size: 24px; font-weight: bold;">Age:  </span>
   * Age is an important factor in predicting diabetes risk. 
   * As individuals get older, their risk of developing diabetes increases. 
   * This is partly due to factors such as reduced physical activity, changes in hormone levels, and a higher likelihood of developing other health conditions that can contribute to diabetes.

# <span style="color:rgb(102, 133, 190); font-size: 24px; font-weight: bold;">Gender:   </span>
   * Gender can play a role in diabetes risk, although the effect may vary. 
   * For example, women with a history of gestational diabetes (diabetes during pregnancy) have a higher risk of developing type 2 diabetes later in life. 
   * Additionally, some studies have suggested that men may have a slightly higher risk of diabetes compared to women.

# <span style="color:rgb(102, 133, 190); font-size: 24px; font-weight: bold;"> Body Mass Index (BMI):  </span>
   * BMI is a measure of body fat based on a person's height and weight. 
   * It is commonly used as an indicator of overall weight status and can be helpful in predicting diabetes risk. 
   * Higher BMI is associated with a greater likelihood of developing type 2 diabetes. Excess body fat, particularly around the waist, can lead to insulin resistance and impair the body's ability to regulate blood sugar levels.

# <span style="color:rgb(102, 133, 190); font-size: 24px; font-weight: bold;">Hypertension:  </span>
   * Hypertension, or high blood pressure, is a condition that often coexists with diabetes. 
   * The two conditions share common risk factors and can contribute to each other's development. 
   * Having hypertension increases the risk of developing type 2 diabetes and vice versa. 
   * Both conditions can have detrimental effects on cardiovascular health.

# <span style="color:rgb(102, 133, 190); font-size: 24px; font-weight: bold;">Heart Disease: </span>
   * Heart disease, including conditions such as coronary artery disease and heart failure, is associated with an increased risk of diabetes. 
   * The relationship between heart disease and diabetes is bidirectional, meaning that having one condition increases the risk of developing the other. 
   * This is because they share many common risk factors, such as obesity, high blood pressure, and high cholesterol.

# <span style="color:rgb(102, 133, 190); font-size: 24px; font-weight: bold;">Smoking History:  </span>
   * Smoking is a modifiable risk factor for diabetes. 
   * Cigarette smoking has been found to increase the risk of developing type 2 diabetes. 
   * Smoking can contribute to insulin resistance and impair glucose metabolism. 
   * Quitting smoking can significantly reduce the risk of developing diabetes and its complications.

# <span style="color:rgb(102, 133, 190); font-size: 24px; font-weight: bold;">HbA1c Level:  </span>
   * HbA1c (glycated hemoglobin) is a measure of the average blood glucose level over the past 2-3 months. 
   * It provides information about long-term blood sugar control. 
   * Higher HbA1c levels indicate poorer glycemic control and are associated with an increased risk of developing diabetes and its complications.

# <span style="color:rgb(102, 133, 190); font-size: 24px; font-weight: bold;">Blood Glucose Level:   </span>
   * Blood glucose level refers to the amount of glucose (sugar) present in the blood at a given time. 
   * Elevated blood glucose levels, particularly in the fasting state or after consuming carbohydrates, can indicate impaired glucose regulation and increase the risk of developing diabetes. 
   * Regular monitoring of blood glucose levels is important in the diagnosis and management of diabetes.

```python
# Define X and target y
X = df.drop(['diabetes'], axis=1)
y = df['diabetes']
```

```python
# Label Encoding for categorical columns
label_encoders = {}
for column in X.select_dtypes(include=['object']).columns:
    le = LabelEncoder()
    X[column] = le.fit_transform(X[column])
    label_encoders[column] = le
```

```python
# Resampling to balance the dataset

def balance_data(X, y, technique='oversample'):
    if technique == 'oversample':
        sampler = RandomOverSampler(random_state=42)
    elif technique == 'undersample':
        sampler = RandomUnderSampler(random_state=42)
    elif technique == 'hybrid':
        sampler = SMOTEENN(random_state=42)
    else:
        raise ValueError("Invalid technique. Choose 'oversample', 'undersample', or 'hybrid'.")
    X_resampled, y_resampled = sampler.fit_resample(X, y)
    return X_resampled, y_resampled

X_resampled, y_resampled = balance_data(X, y, technique='hybrid')
```

```python
# spliting the data into train and test 
X_train, X_test, y_train, y_test = train_test_split(X_resampled, y_resampled, test_size=0.2, random_state=42)
```

```python
# Initialize StandardScaler
scaler = MinMaxScaler()

# Fit scaler on training data and transform both training and test sets
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

  <h1 style="color: #ECF0F1; font-size: 30px; font-weight: bold; text-align: center; text-shadow: 2px 2px 4px #000000;">

```python
# the model
xgb_model = XGBClassifier(random_state = 42)  # Suppress warnings
xgb_model.fit(X_train_scaled, y_train)
```

```python
model_LR = LogisticRegression()
model_LR.fit(X_train_scaled, y_train)
```

```python
# Initialize the Random Forest Classifier
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model.fit(X_train_scaled, y_train)
```

1. Precision:

    * Represents the proportion of true positive predictions for a given class out of all predictions made for that class.
    * Formula:
    $$
    \text{Precision} = \frac{\text{True Positives (TP)}}{\text{True Positives (TP)} + \text{False Positives (FP)}}
    $$

    * Interpretation:
        * Precision measures how reliable the model is when it predicts a sample as belonging to a particular class.
2. Recall:    
    * Represents the proportion of true positive predictions for a given class out of all actual instances of that class in the dataset.
    * Formula:
    $$
    \text{Recall} = \frac{\text{True Positives (TP)}}{\text{True Positives (TP)} + \text{False Negatives (FN)}}
    $$
    
    * Interpretation:
        * Recall measures how well the model identifies all the true instances of a given class.

3. f1-score:  
    * Represents the harmonic mean of precision and recall, providing a balanced measure of the model's accuracy
    * Formula:
    $$
    \text{F1-Score} = 2 \times \frac{\text{Precision} \times \text{Recall}}{\text{Precision} + \text{Recall}}
    $$
    * Interpretation:
        * The F1-score is useful for evaluating models where there is an imbalance between precision and recall. It combines both into a single metric.

4. support:   
    * The number of actual occurrences of each class in the dataset. It represents the true count of each class.
    * Formula:
$$
\text{Accuracy} = \frac{\text{Number of Correct Predictions}}{\text{Total Number of Predictions}}
$$


```python
models = [xgb_model, model_LR, rf_model]
models_names = ["XGBClassifier", 'LogisticRegression', 'RandomForestClassifier']
result1 = []  # For accuracy
result2 = []  # For recall
result3 = []  # For F1-score

for i in range(len(models)):
    # Predicting with each model
    y_pred = models[i].predict(X_test_scaled)
    
    # Calculating accuracy
    accuracy = accuracy_score(y_test, y_pred)
    print(f"The accuracy of the {models_names[i]} : {accuracy * 100:.2f} %\n")
    
    # Classification report
    print(f"Classification Report of {models_names[i]}:")
    print(classification_report(y_test, y_pred))

    # Confusion matrix
    conf_matrix = confusion_matrix(y_test, y_pred)
    
    # Recall and F1-score
    recall = recall_score(y_test, y_pred)
    f1 = f1_score(y_test, y_pred)

    # Storing results for further analysis
    result1.append(accuracy)
    result2.append(recall)
    result3.append(f1)

    # Plot the heatmap of the confusion matrix
    plt.figure(figsize=(8, 6))
    sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues', cbar=True, 
                xticklabels=["No Diabetes", "Diabetes"], yticklabels=["No Diabetes", "Diabetes"])
    plt.title(f"Confusion Matrix Heatmap of {models_names[i]}")
    plt.xlabel("Predicted Labels")
    plt.ylabel("True Labels")
    plt.show()

    print("__________________________________________________________________________________________________________________\n")
```

```python
final_result = pd.DataFrame({'Algorithm':models_names, 
              'Accuracies':result1,
              'Recall':result2,
              'F1_Score':result3
              })

fig, ax = plt.subplots(figsize=(20,5))

plt.plot(final_result['Algorithm'], result1, label='Acurracies')
plt.plot(final_result['Algorithm'], result2, label='Recall')
plt.plot(final_result['Algorithm'], result3, label='F1_Score')

plt.legend()
plt.show()
```
